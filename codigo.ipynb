### **4. Limpeza de Dados**

```python
# Quantidade antes
print(df.shape)

# Removendo duplicados
df = df.drop_duplicates()

# Quantidade depois
print(df.shape)
```

```python
# Método IQR
Q1 = df[['Unit price','Quantity','Total']].quantile(0.25)
Q3 = df[['Unit price','Quantity','Total']].quantile(0.75)
IQR = Q3 - Q1

# Removendo outliers
df = df[~((df[['Unit price','Quantity','Total']] < (Q1 - 1.5 * IQR)) |
          (df[['Unit price','Quantity','Total']] > (Q3 + 1.5 * IQR))).any(axis=1)]

# Boxplot das variáveis numéricas após a remoção de dados
plt.figure(figsize=(14,6))
sns.boxplot(data=df[['Unit price','Quantity','Tax 5%','Total','Rating','cogs','gross margin percentage','gross income']])
plt.title('Variáveis Numéricas')
plt.show()
```

##### **Análise da Limpeza de Dados**
Não há registros duplicados nem necessidade de ajuste de tipos.

Foi realizado tratamento de outliers utilizando o método IQR, reduzindo a influência de compras extremamente fora do padrão.

O gráfico 5 de mapa de calor, permite identificar relações entre variáveis numéricas. A variável “Total” possui forte correlação positiva com “Quantity” e “Unit price”.

### **8. Relações Entre Variáveis**

```python
# Gráfico 8: Linha de tendência
plt.figure(figsize=(8,5))
sns.regplot(x='Quantity', y='Total', data=df, color='orange', line_kws={'color':'red'})
plt.title('Correlação entre Quantidade e Valor Total', fontsize=12, fontweight='bold', color="black")
plt.xlabel('Quantidade')
plt.ylabel('Valor Total')
plt.savefig('08_gráfico.png', bbox_inches='tight')
plt.show()
```

##### **Innterpretação**
O gráfico 8 mostra que existe forte correlação positiva entre quantidade comprada e valor total das vendas.

Isso indica que compras com maior quantidade de produtos tendem a gerar maior faturamento.
